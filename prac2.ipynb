операторов реализуйте игру «FizzBuzz» с тремя условиями: кратно 3 — «Fizz»,

1. В чём принципиальное отличие циклов for и while в Python?
Приведите по одному примеру задачи, где каждый из них будет наиболее уместен,
и кратко обоснуйте выбор.

Цикл for используется, когда известно, сколько раз нужно повторить действия
или есть итерируемый объект (список, строка, range).
Цикл while используется, когда повторение зависит от условия, и заранее
неизвестно, сколько итераций потребуется.
Пример для for

```
total = 0
    for i in range(1, 101):
        total += i
```

Пример для while

```
password = ""
    while password != "secret":
        password = input("Введите пароль: ")
```

2. Как работает блок else внутри цикла (for/while)? Опишите, при каких
условиях он выполняется, а при каких – нет. Приведите два коротких примера кода:
в одном else срабатывает, в другом – нет (из за break).

Else после for или while выполняется, если цикл завершился нормально,
то есть без break. Если цикл был прерван через break, else не выполняется.
Пример, где else срабатывает

```
for i in range(3):
        print(i)
    else:
        print("Цикл завершён без break")
```

Пример, где else не срабатывает
```
for i in range(5):
        if i == 3:
            break
    else:
        print("Это не выполнится")
```

3. Объясните разницу между операторами break и continue. Напишите
фрагмент кода с циклом while, где используются оба оператора, и
прокомментируйте, что произойдёт на каждой итерации.

Оператор break полностью прерывает цикл.
Оператор continue пропускает оставшуюся часть текущей итерации и переходит к следующей.
```
i = 0
while i < 5:
  i += 1
    if i == 2:
      continue  # пропустить вывод числа 2
    if i == 4:
      break     # полностью выйти из цикла
print(i)
```

4. Какие типичные ошибки возникают при организации циклов в Python?

пример ошибочного кода и его исправленную версию.

1) Бесконечный цикл — не изменяется условие выхода.

Ошибка:
```
i = 0
while i < 5:
  print(i)
```
Исправление:
```
i = 0
while i < 5:
  print(i)
  i += 1
```
2) Неверный диапазон range — верхняя граница не включается.

Ошибка: числа 1..4, а не 1..5
```
for i in range(1, 5):
print(i)
```
Исправление:
```
for i in range(1, 6):
print(i)
```

3) Неправильное условие выхода — например, сравнение вместо присваивания.

Ошибка:
```
       x = 0
       while x = 10:
           x += 1
```
Исправление:
```
       x = 0
       while x < 10:
           x += 1
```

С циклом:
```
nums = [5, 2, 8]

  for n in nums:
    total += n
      if n < minimum:
        minimum = n
```

Со встроенными функциями:
